```python
def dobro(n):
    return n * 2

print(dobro(5))   # 10
print(dobro(3))   # 6
```

```python
def area_triangulo(base, altura):
    return (base * altura) / 2

print(area_triangulo(4, 3))  # 6.0
```

```python
def maior_valor(lista):
    maior = lista[0]
    for n in lista:
        if n > maior:
            maior = n
    return maior

print(maior_valor([3, 7, 1, 9, 2]))  # 9
```

```python
lista = []

for i in range(10):
    n = int(input(f"Número {i+1}: "))
    if n > 50:
        lista.append(n)

print("Maiores que 50:", lista)
```

```python
frase = input("Digite uma frase: ")
vogais = "aeiouAEIOU"
contador = 0

for letra in frase:
    if letra in vogais:
        contador += 1

print("Vogais:", contador)
```
